```python
import pandas as pd
from mbiomekit.machine_learning import random_forest
```

```python
ft = pd.read_csv('data/ft_genus_prv01.tsv', sep='\t', header=0, index_col=0)
lbs = pd.read_csv('data/group.tsv', sep='\t', header=0, index_col=0)['group1']
```

```python
prd = random_forest.RandomForestPredictor()
prd.load_data(ft.T, lbs)
prd.set_random_forest_parameters()
prd.set_cross_validation_parameters()
prd.cross_validate_models()
prd.train_final_model()  
prd.save_results('data/3-results/machine_learning/results')
```

```python
# load enterotypes 
ent = pd.read_csv('data/enterotype.tsv', sep='\t', header=0, index_col=0)
for c in [0, 1, 2, 3]:
    ids = ent.index[ent['cluster'] == c]
    print(len(ids))
    
    prd = random_forest.RandomForestPredictor()
    prd.load_data(ft.T.loc[ids, :], lbs[ids])
    prd.set_random_forest_parameters()
    prd.set_cross_validation_parameters()
    prd.cross_validate_models()
    prd.train_final_model()  
    prd.save_results(f'data/3-results/machine_learning/results_cl{c}')
    
```
