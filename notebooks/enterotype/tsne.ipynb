```python
# Load libraries
import pandas as pd 
from mbiomekit.beta_diversity import tsne
```

```python
# Load the distance matrix (weighted UniFrac)
dmat = pd.read_csv('data/2-processed/amplicon/diversity/beta_distance/weighted_unifrac_distance/distance-matrix.tsv', sep='\t', header=0, index_col=0)
```

```python
# Calculate t-SNE and clustering
ta = tsne.TSNEAnaysis()
ta.load_distance_matrix(dmat)
ta.tsne()
ta.choose_cluster_model(fout='data/3-results/enterotype/aic_bic.tsv')
ta.cluster(n_clusters=4, fout='data/3-results/enterotype/tsne_cluster.tsv')
```

```python
# Plotting
tp = tsne.TSNEPlotter()
tp.plot_aic_bic('data/3-results/enterotype/aic_bic.tsv', fout='figures/enterotype/aic_bic.pdf')
tp.plot_tsne_cluster('data/3-results/enterotype/tsne_cluster.tsv', fout='figures/enterotype/tsne.pdf', palette={0: '#326a8190' , 1: '#7876b190', 2: '#9fbcca90', 3: '#aaadb290'})
```
