```python
# Load libraries
import pandas as pd
from mbiomekit.alpha_diversity import plotter
```

```python
# Load alpha diversity data
ad = pd.read_csv('data/alpha_diversity.tsv', sep='\t', header=0, index_col=0)
gr = pd.read_csv('data/group.tsv', sep='\t', header=0, index_col=0)['group1']
```

```python
# Plot 
## Set color palette
palette = {'Healthy': '#aaadb2',
           'Prevent': '#9fbcca',
           'Predict': '#326a81'}

## Plot alpha diversity
for d in ['faith_pd', 'chao1']:
    pl = plotter.AlphaDiversityPlot()
    pl.load_alpha_diversity(ad[d])
    pl.load_group(gr)
    pl.plot_by_group(group_order=['Healthy', 'Prevent', 'Predict'], palette=palette, fout=f'figures/alpha_diversity/{d}.pdf')
```
