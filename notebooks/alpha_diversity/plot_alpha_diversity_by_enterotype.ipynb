```python
# Load libraries
import pandas as pd
from mbiomekit.alpha_diversity import plotter

```

```python
# Set parameters
palette = {'Healthy': '#aaadb2',
           'Prevent': '#9fbcca',
           'Predict': '#326a81'}
```

```python
# Load data files
ad = pd.read_csv('data/alpha_diversity.tsv', sep='\t', header=0, index_col=0)
gr = pd.read_csv('data/group.tsv', sep='\t', header=0, index_col=0)['group1']
ent = pd.read_csv('data/enterotype.tsv', sep='\t', header=0, index_col=0)
```

```python
# Group by enterotype 
egrp = {}
for id, val in ent.iterrows():
    val = str(val['cluster'])
    if val not in egrp:
        egrp[val] = [id]
    else:
        egrp[val].append(id)

```

```python
# Plot alpha diversity 

palette = {'Healthy': '#aaadb2',
           'Prevent': '#9fbcca',
           'Predict': '#326a81'}


for g in egrp:
    for d in ['faith_pd']:
        pl = plotter.AlphaDiversityPlot()
        print(egrp[g])
        pl.load_alpha_diversity(ad.loc[egrp[g], d])
        pl.load_group(gr.loc[egrp[g]])
        pl.plot_by_group(group_order=['Healthy', 'Prevent', 'Predict'], palette=palette, fout=f'figures/alpha_diversity/{d}_{g}.pdf', ylim=[3.5, 30])
```
